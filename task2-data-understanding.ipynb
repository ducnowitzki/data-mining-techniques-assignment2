```python
group = 'position'
targets = list(set(data_df.columns) - set(['srch_id', 'date_time', 'site_id', 'visitor_location_country_id', 'prop_country_id', 'prop_id', 'srch_destination_id', 'click_bool', 'booking_bool', 'gross_bookings_usd', 'position']))
```

```python
for target in targets:
    mean_data = data_df.groupby(group)[target].mean().reset_index()
    median_data = data_df.groupby(group)[target].median().reset_index()

    # Setting up the subplot grid
    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 6))

    # Plotting Mean
    axes[0].plot(mean_data[group], mean_data[target], marker='o', color='blue')
    axes[0].set_title('mean ' + target + ' per ' + group)
    axes[0].set_xlabel(group)
    axes[0].set_ylabel('mean ' + target)
    axes[0].grid(True)

    # Plotting Median
    axes[1].plot(median_data[group], median_data[target], marker='o', color='green')
    axes[1].set_title('median ' + target + ' per ' + group)
    axes[1].set_xlabel(group)
    axes[1].set_ylabel('median ' + target)
    axes[1].grid(True)

    # Show the plots
    plt.tight_layout()

    file_name = 'figs/' + group + '/' + target + '_per_' + group + '.png'
    plt.savefig(file_name)
    # plt.show()
```
